```python
import time
import datetime
from dataclasses import dataclass
from ifm_konwoledge_enginnering.Block_World.Project.src.search_path.common import SearchResult
from ifm_konwoledge_enginnering.Block_World.Project.src.utils.state_generator import start_goal_generator
from ifm_konwoledge_enginnering.Block_World.Project.src.utils.path_utils import check_path
from ifm_konwoledge_enginnering.Block_World.Project.src.search_path.dfs import dfs, dfs_with_stats

#TODO: Zadanie 4. Wariant podstawowy (generacja i czasy dla DFS)
@dataclass
class Result:
    name: str
    n_blocks: int
    search_result: SearchResult
    time:float


def test_n_block_dsf(n_block:int, seed:int=42):
    '''
    Generate random start and goal, run dfs, return path and time.
    '''

    start_goal_obj = start_goal_generator(n_block, seed)
    start, goal = start_goal_obj.start, start_goal_obj.goal

    print("Start DFS for:\n"
          f"start: {start.placement}\n"
          f"goal: {goal.placement}\n")

    start_t = time.time()
    search_result = dfs_with_stats(start, goal)
    stop_t = time.time()

    total_time = stop_t - start_t


    print("Checking if path is correct")
    if check_path(start, search_result.path):
        print("Path is correct")
    else:
        ValueError("Path is not correct")


    return Result(
        name=f"DFS_{n_block}_seed={seed}_{datetime.datetime.now().strftime('%Y%m%d-%H%M%S')}",
        n_blocks=n_block,
        search_result=search_result,
        time=total_time,
    )
```
