# Laboratorium 2 — świat Wumpusa i wnioskowanie w ASP (Clingo)

## Problem

*Świat Wumpusa* (Russell & Norvig, *Artificial Intelligence: A Modern Approach*, rozdz. 7) to klasyczny przykład środowiska dla **agenta opartego na wiedzy**. Jaskinia jest prostokątną planszą pól $(x, y)$, $1 \le x \le W$, $1 \le y \le H$. Myśliwy zaczyna na polu wejściowym $(1,1)$, patrząc na wschód. W jaskini znajdują się:

* **jamy** — wejście do jamy oznacza śmierć,
* **wumpus** — potwór, który zjada myśliwego wchodzącego na jego pole (martwy wumpus jest już niegroźny),
* **złoto** — cel wyprawy.

Pole wejściowe jest zawsze bezpieczne. Agent **nie widzi** planszy — zna tylko jej rozmiar oraz to, co poczuje na polach, które odwiedził:

| bodziec | znaczenie |
|---|---|
| *Stench* (smród) | na polu sąsiednim (krawędzią) jest wumpus — także martwy |
| *Breeze* (powiew) | na polu sąsiednim jest jama |
| *Glitter* (blask) | na bieżącym polu leży złoto |
| *Bump* (uderzenie) | agent wszedł w ścianę |
| *Scream* (krzyk) | właśnie wystrzelona strzała zabiła wumpusa |

**Akcje:** `left`, `right` (obrót o 90°), `forward` (krok), `grab` (podniesienie złota), `shoot` (strzała leci prosto do ściany i zabija pierwszego napotkanego żywego wumpusa), `climb` (wyjście z jaskini, tylko na polu $(1,1)$).

**Punktacja (jak w AIMA):** $+1000$ za każde wyniesione złoto, $-1000$ za śmierć, $-10$ za strzałę, $-1$ za każdą akcję.

## Podejście

Agent zapisuje w bazie wiedzy wyłącznie obserwacje (odwiedzone pola, percepcje, strzały). Reguły świata zapisane są w programie ASP `wumpus_kb.lp`: każdy model stabilny to jeden możliwy układ ukrytej części jaskini zgodny z obserwacjami. Clingo liczy

* **konsekwencje ostrożne** (prawdziwe we wszystkich modelach) — to, co agent **wie** (np. „pole jest bezpieczne”, „tu na pewno jest jama”),
* **konsekwencje odważne** (prawdziwe w jakimś modelu) — to, co jest **możliwe**.

Na tej podstawie agent wybiera cel (zgodnie z priorytetem: podnieś złoto → wróć z całym złotem → idź na najbliższe pewnie bezpieczne pole → zastrzel ewentualnego wumpusa na polu bez jamy → [opcjonalnie] zaryzykuj → wyjdź z jaskini), a trasę wyznacza przeszukiwaniem wszerz po polach, o których wie, że są bezpieczne.

Pliki: `world.py` (środowisko), `inference.py` (baza wiedzy i wywołania Clingo), `agent.py` (strategia i planowanie), `simulate.py` (pętla gry i statystyki).

## Rozważane warianty

Funkcja `run_game(config, max_actions, risky, agent_knows_gold, verbose)` pozwala porównywać następujące warianty:

1. **Standardowy świat z AIMA** (`config = AIMA_WORLD`) — plansza 4×4 z rys. 7.2: jamy w $(3,1)$, $(3,3)$, $(4,4)$, wumpus w $(1,3)$, złoto w $(2,3)$.
2. **Losowe światy** (`generate_world(width, height, pit_prob, n_wumpus, n_gold, rng=...)`):
   * dowolny rozmiar planszy (także prostokątnej),
   * każde pole poza wejściem jest jamą niezależnie z prawdopodobieństwem `pit_prob` (domyślnie 0.2),
   * **wiele wumpusów** — agent ma tyle strzał, ilu jest wumpusów; wumpus może stać na jamie,
   * **wiele sztuk złota** — złoto leży tylko na polach bez jam.

   Losowy świat może być nierozwiązywalny (np. złoto otoczone jamami albo brak bezpiecznego ruchu już na starcie) — wtedy rozsądny agent po prostu wychodzi z jaskini.
3. **Znana / nieznana liczba sztuk złota** (`agent_knows_gold`) — jeśli agent wie, ile jest złota, wraca od razu po zebraniu wszystkiego; w przeciwnym razie eksploruje, dopóki istnieją pewnie bezpieczne pola.
4. **Agent ostrożny / ryzykujący** (`risky`) — agent ostrożny wchodzi tylko na pola, których bezpieczeństwo **wynika logicznie** z wiedzy; ryzykujący, gdy takich pól brak, wchodzi na najmniej podejrzane pole, które nie jest na pewno śmiertelne (zamiast od razu się wycofać).

# Porównanie Trzech Agnetów
Nie ryzykujący -> w pyrzpadku ryzykwa wraca

Ryzykujący -> wybiera pierwsze miejsce, które napewno nie jest śmiertelne

Ryzykujący z prawdopodobińtwe + treshold -> wybiera najmniejsze ryzyko mniejsze niż $0.5$

```python
from dataclasses import astuple
from world import generate_world
import numpy as np

@dataclass
class WorldParams:
    width: int
    height: int
    pit_prob: float
    n_wumpus: int
    n_gold: int
    gold_int_pit: bool
    rng: random.Random

    def __iter__(self):
        return iter(astuple(self))


def run(risky:bool, uses_probability:bool,  world_params: WorldParams, max_actions:bool=1000, unknown_gold:bool = False) -> int:
    world = generate_world(*list(world_params))

    gr = run_game(
        config=world,
        max_actions=max_actions,
        risky=risky,
        agent_knows_gold = True,
        verbose=verbose,
        pit_prob=world_params.pit_prob,
        use_danger_probability=uses_probability)

    return gr.score

N = 100
world_params = WorldParams(
    width=4,
    height=4,
    pit_prob=0.2,
    n_wumpus=1,
    n_gold=1,
    gold_int_pit=False,
    rng=random.Random(),
)

risky_scores_not_prob = np.zeros(N)
risky_prob = np.zeros(N)
safe_scores = np.zeros(N)
for i in range(N):
    world_params.rng.seed(i)
    risky_scores_not_prob[i] = run(True, False, world_params)
    risky_prob[i]  = run(True, True, world_params)
    safe_scores[i] = run(False, False, world_params)

```

## Wyniki eksperymentu — porównanie strategii

Poniższe wykresy porównują wyniki agenta ostrożnego, ryzykownego oraz ryzykującego z rankingiem prawdopodobieństwa zagrożenia. Strategie testowano na tych samych ziarnach losowych, więc każda z nich otrzymała ten sam zestaw światów. Wykresy pokazują rozkład wyników, ich częstość oraz średnią z błędem standardowym.


```python
print(f'risky_scores with probability: {risky_prob.mean()}\n'
      f'risky_score without probability: {risky_scores_not_prob.mean()}\n'
      f'safe_scores: {safe_scores.mean()}\n')
```

```python
import matplotlib.pyplot as plt

strategies = [
    ("Agent ostrożny", np.asarray(safe_scores)),
    ("Ryzykowny", np.asarray(risky_scores_not_prob)),
    ("Ryzykowny + zliczanie", np.asarray(risky_prob)),
]
colors = ["#4C78A8", "#F58518", "#54A24B"]
labels = [name for name, _ in strategies]
values = [scores for _, scores in strategies]

fig = plt.figure(figsize=(17, 9), constrained_layout=True)
grid = fig.add_gridspec(2, 3, height_ratios=(1, 1.2))
box_ax = fig.add_subplot(grid[0, :2])
mean_ax = fig.add_subplot(grid[0, 2])
hist_axes = [fig.add_subplot(grid[1, column]) for column in range(3)]
fig.suptitle(
    f"Porównanie strategii w {N} losowych światach "
    f"(prior jamy: {world_params.pit_prob:.0%})",
    fontsize=15,
)

# Score distributions: median, quartiles, and mean.
boxplot = box_ax.boxplot(
    values,
    tick_labels=labels,
    patch_artist=True,
    showmeans=True,
)
for patch, color in zip(boxplot["boxes"], colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)
box_ax.set_title("Rozkład wyników")
box_ax.set_ylabel("Wynik gry (punkty)")
box_ax.tick_params(axis="x", rotation=8)
box_ax.grid(axis="y", alpha=0.25)

# Separate histograms avoid overlap while sharing identical bins and scales.
all_scores = np.concatenate(values)
bins = np.histogram_bin_edges(all_scores, bins="auto")
score_min, score_max = all_scores.min(), all_scores.max()
score_padding = max((score_max - score_min) * 0.04, 1)
max_bin_count = max(np.histogram(scores, bins=bins)[0].max() for scores in values)
for ax, (name, scores), color in zip(hist_axes, strategies, colors):
    ax.hist(scores, bins=bins, color=color, alpha=0.8, edgecolor="white")
    ax.set_title(name)
    ax.set_xlabel("Wynik gry (punkty)")
    ax.set_xlim(score_min - score_padding, score_max + score_padding)
    ax.set_ylim(0, max_bin_count * 1.08)
    ax.grid(axis="y", alpha=0.25)
hist_axes[0].set_ylabel("Liczba gier")

# Mean score with standard error across the sampled worlds.
means = [scores.mean() for scores in values]
standard_errors = [
    scores.std(ddof=1) / np.sqrt(len(scores)) if len(scores) > 1 else 0.0
    for scores in values
]
short_labels = ["Ostrożny", "Ryzykowny", "Ryzyk. + zlicz."]
bars = mean_ax.bar(
    short_labels,
    means,
    yerr=standard_errors,
    capsize=5,
    color=colors,
    alpha=0.85,
)
mean_ax.bar_label(bars, fmt="%.0f", padding=3)
mean_ax.axhline(0, color="black", linewidth=0.8)
mean_ax.set_title("Średni wynik (± błąd standardowy)")
mean_ax.set_ylabel("Średni wynik (punkty)")
mean_ax.tick_params(axis="x", labelsize=9, rotation=10)
mean_ax.grid(axis="y", alpha=0.25)

plt.show()
```
